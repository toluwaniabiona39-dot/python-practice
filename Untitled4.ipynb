```python
customer = "Toluwani"

rice_price = 1500
rice_qty = 3
rice_total = rice_price * rice_qty

bread_price = 350
bread_qty = 5
bread_total = bread_price * bread_qty

milk_price = 200
milk_qty = 3
milk_total = milk_price * milk_qty

subtotal = rice_total + bread_total + milk_total
vat = subtotal * 0.075
discount = (subtotal + vat) * 0.10
total = subtotal + vat - discount

print("----- RECEIPT -----")
print("Customer:", customer)
print("Rice total:", rice_total)
print("Bread total:", bread_total)
print("Milk total:", milk_total)
print("Subtotal:", subtotal)
print("VAT (7.5%):", round(vat, 2))
print("Discount (10%):", round(discount, 2))
print("Total to pay:", round(total, 2))
print("-------------------")

amount_paid = 7000
change = amount_paid - total
print("Amount paid:", amount_paid)
print("Change:", round(change, 2))
```
